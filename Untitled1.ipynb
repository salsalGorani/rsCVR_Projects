```python
# DEFs

def natural_key(path):
    """
    파일명에 포함된 숫자를 기준으로 자연 정렬.
    swra..._00001.nii, swra..._00002.nii 순서 보장용.
    """
    name = Path(path).name
    return [
        int(text) if text.isdigit() else text.lower()
        for text in re.split(r"(\d+)", name)
    ]


def get_volume_index(path):
    """
    파일명 마지막의 _00001.nii 같은 volume index를 추출.
    실패하면 None 반환.
    """
    name = Path(path).name
    m = re.search(r"_(\d+)\.nii(\.gz)?$", name)
    if m:
        return int(m.group(1))
    return None


def find_one(directory, pattern, description):
    files = sorted(directory.glob(pattern), key=natural_key)

    if len(files) == 0:
        raise FileNotFoundError(f"{description} 파일을 찾지 못함: {directory / pattern}")

    if len(files) > 1:
        print(f"[주의] {description} 후보가 여러 개 발견됨. 첫 번째 파일을 사용함.")
        for f in files:
            print("  -", f)

    return files[0]


def load_nifti(path):
    return nib.load(str(path))


def same_grid(img1, img2, affine_tol=1e-4):
    same_shape = img1.shape[:3] == img2.shape[:3]
    same_affine = np.allclose(img1.affine, img2.affine, atol=affine_tol)
    return same_shape and same_affine


def describe_img(img):
    zooms = img.header.get_zooms()
    return {
        "shape": img.shape,
        "zooms": zooms,
        "affine": img.affine,
    }


def save_mask(mask_bool, ref_img, out_path):
    mask_img = nib.Nifti1Image(
        mask_bool.astype(np.uint8),
        ref_img.affine,
        ref_img.header.copy()
    )
    mask_img.set_data_dtype(np.uint8)
    nib.save(mask_img, str(out_path))


def make_4d_from_3d_files(files, out_path, tr):
    """
    3D NIfTI 파일들을 시간축으로 묶어 4D NIfTI 생성.
    원본 파일은 수정하지 않음.
    """
    first_img = load_nifti(files[0])
    first_shape = first_img.shape[:3]
    first_affine = first_img.affine

    n_vol = len(files)
    data_4d = np.empty(first_shape + (n_vol,), dtype=np.float32)

    for i, f in enumerate(files):
        img = load_nifti(f)

        if img.shape[:3] != first_shape:
            raise ValueError(
                f"Shape mismatch at {f}\n"
                f"Expected: {first_shape}, got: {img.shape[:3]}"
            )

        if not np.allclose(img.affine, first_affine, atol=1e-4):
            raise ValueError(
                f"Affine mismatch at {f}\n"
                "affine MUST be same between swra images"
            )

        data_4d[..., i] = np.asarray(img.dataobj, dtype=np.float32)

    # Create blank 4D NIfTI
    out_img = nib.Nifti1Image(data_4d, first_affine)

    # set dtype
    out_img.header.set_data_dtype(np.float32)

    # Set voxel size + TR (4D)
    spatial_zooms = first_img.header.get_zooms()[:3]
    out_img.header.set_zooms(spatial_zooms + (tr,))

    # qform/sform code conversion to Python int
    qform_code = int(np.asarray(first_img.header["qform_code"]))
    sform_code = int(np.asarray(first_img.header["sform_code"]))

    # Set aligned=2 :: code가 0이면 unknown임
    if qform_code == 0:
        qform_code = 2
    if sform_code == 0:
        sform_code = 2

    out_img.set_qform(first_affine, code=qform_code)
    out_img.set_sform(first_affine, code=sform_code)

    nib.save(out_img, str(out_path))

    return out_img


def resample_prob_to_ref(prob_img, ref_img, order=1):
    """
    segmentation probability map을 BOLD grid로 resampling.
    order=1: probability map에 적절한 linear interpolation.
    """
    return resample_from_to(
        prob_img,
        (ref_img.shape[:3], ref_img.affine),
        order=order
    )


def make_global_regressor(data_4d, mask_bool, scaling="percent"):
    """
    4D BOLD에서 mask 내부 global mean regressor 생성.
    """
    if data_4d.ndim != 4:
        raise ValueError("data_4d는 4D array여야 함.")

    if mask_bool.shape != data_4d.shape[:3]:
        raise ValueError(
            f"Mask shape mismatch. mask={mask_bool.shape}, bold={data_4d.shape[:3]}"
        )

    ts = data_4d[mask_bool, :]
    global_mean = np.nanmean(ts, axis=0)

    if scaling == "raw":
        reg = global_mean

    elif scaling == "percent":
        baseline = np.nanmean(global_mean)
        if baseline == 0:
            raise ValueError("Global mean baseline이 0이므로 percent scaling 불가.")
        reg = 100.0 * (global_mean / baseline - 1.0)

    elif scaling == "zscore":
        mu = np.nanmean(global_mean)
        sd = np.nanstd(global_mean)
        if sd == 0:
            raise ValueError("Global mean std가 0이므로 z-score scaling 불가.")
        reg = (global_mean - mu) / sd

    else:
        raise ValueError("scaling은 'raw', 'percent', 'zscore' 중 하나여야 함.")

    return reg.astype(np.float64), global_mean.astype(np.float64)


def run_command(cmd, log_path):
    """
    shell command 실행 후 stdout/stderr를 log 파일에 저장.
    """
    print("\n[Command]")
    print(" ".join(map(str, cmd)))
    print()

    with open(log_path, "w") as f:
        f.write(" ".join(map(str, cmd)) + "\n\n")
        proc = subprocess.run(
            cmd,
            stdout=f,
            stderr=subprocess.STDOUT,
            text=True
        )

    if proc.returncode != 0:
        raise RuntimeError(
            f"Command failed with return code {proc.returncode}\n"
            f"Log file: {log_path}"
        )

    return proc.returncode
```

```python
# MASK :: Load segmentation maps and combined masks
print("\n[Step] Making masks")

gm_img = load_nifti(gm_path)
wm_img = load_nifti(wm_path)
csf_img = load_nifti(csf_path)

# segmentation map끼리 grid 확인
if not same_grid(gm_img, wm_img):
    raise ValueError("GM과 WM segmentation grid가 다름.")
if not same_grid(gm_img, csf_img):
    raise ValueError("GM과 CSF segmentation grid가 다름.")

# BOLD grid와 segmentation grid가 다르면 BOLD grid로 resampling
if same_grid(gm_img, bold4d_img):
    print("Segmentation grid == BOLD grid. Resampling 생략.")
    gm_ref_img = gm_img
    wm_ref_img = wm_img
    csf_ref_img = csf_img

else:
    print("[info] Segmentation grid ≠ BOLD grid")
    print("       Resampling: Segmentation probability maps --> BOLD grid")

    if not RESAMPLE_SEG_TO_BOLD:
        raise ValueError(
            "Matching FAILED: Segmentation grid ≠ BOLD grid "
            "Make 'RESAMPLE_SEG_TO_BOLD=True' to make rapidtide mask"
        )

    gm_ref_img = resample_prob_to_ref(gm_img, bold4d_img, order=1)
    wm_ref_img = resample_prob_to_ref(wm_img, bold4d_img, order=1)
    csf_ref_img = resample_prob_to_ref(csf_img, bold4d_img, order=1)

gm = gm_ref_img.get_fdata(dtype=np.float32)
wm = wm_ref_img.get_fdata(dtype=np.float32)
csf = csf_ref_img.get_fdata(dtype=np.float32)

# MASK COMBINATIONs, Thredholds
brain_mask = (gm + wm + csf) > 0.5
matter_mask = (gm + wm) > 0.5


brain_mask_path = out_dir / f"{SUBJECT}_desc-brain_mask.nii.gz"
save_mask(brain_mask, bold4d_img, brain_mask_path)
print("Saved:", brain_mask_path)
print("brain_mask voxels :", int(brain_mask.sum()))

matter_mask_path = out_dir / f"{SUBJECT}_desc-matter_mask.nii.gz"
save_mask(matter_mask, bold4d_img, matter_mask_path)
print("Saved:", matter_mask_path)
print("matter_mask voxels:", int(matter_mask.sum()))
```

```python
# Load background BOLD
bold_img = nib.load(str(bold4d_path))
bold_data = bold_img.get_fdata(dtype=np.float32)
nx, ny, nz, nt = bold_data.shape
print("BOLD shape:", bold_data.shape)

# Collect rapidtide NIfTI maps
rt_nii_files = sorted(
    list(rt_result_dir.rglob("*.nii")) + list(rt_result_dir.rglob("*.nii.gz")),
    key=lambda x: x.name
)

if len(rt_nii_files) == 0:
    raise FileNotFoundError(f"No NIfTI files found in: {rt_result_dir}")

rt_map_dict = {f.name: f for f in rt_nii_files}

print(f"Number of rapidtide maps: {len(rt_map_dict)}")
for name in rt_map_dict:
    print(name)

# Def for load 3D map
def load_rt_map(map_name):
    path = rt_map_dict[map_name]
    img = nib.load(str(path))
    data = img.get_fdata(dtype=np.float32)

    # rapidtide 결과가 4D singleton 형태일 경우 처리
    data = np.squeeze(data)

    if data.ndim != 3:
        raise ValueError(
            f"{map_name} is not a 3D map after squeeze. Shape = {data.shape}"
        )

    if data.shape != bold_data.shape[:3]:
        raise ValueError(
            f"Shape mismatch:\n"
            f"BOLD shape = {bold_data.shape[:3]}\n"
            f"Map shape  = {data.shape}\n"
            f"Map file   = {path}"
        )

    return data

# Interactive rapidtide map viewer
def show_rapidtide_map(
    map_name,
    volume=1,
    plane="axial",
    slice_idx=None,
    alpha=0.65,
    bg_vmin_pct=1,
    bg_vmax_pct=99,
    map_pct=98,
    symmetric=True,
    show_zero_as_transparent=True
):
    # background BOLD
    vol_idx = volume - 1
    bg = bold_data[..., vol_idx]

    # rapidtide map
    rt_map = load_rt_map(map_name)

    if slice_idx is None:
        if plane == "axial":
            slice_idx = nz // 2
        elif plane == "coronal":
            slice_idx = ny // 2
        elif plane == "sagittal":
            slice_idx = nx // 2

    if plane == "axial":
        bg2d = bg[:, :, slice_idx].T
        map2d = rt_map[:, :, slice_idx].T
        label = f"Axial z={slice_idx}"

    elif plane == "coronal":
        bg2d = bg[:, slice_idx, :].T
        map2d = rt_map[:, slice_idx, :].T
        label = f"Coronal y={slice_idx}"

    elif plane == "sagittal":
        bg2d = bg[slice_idx, :, :].T
        map2d = rt_map[slice_idx, :, :].T
        label = f"Sagittal x={slice_idx}"

    else:
        raise ValueError("plane must be axial, coronal, or sagittal")

    # background window
    finite_bg = bg[np.isfinite(bg)]
    bg_vmin, bg_vmax = np.percentile(finite_bg, [bg_vmin_pct, bg_vmax_pct])

    # map window
    finite_map = rt_map[np.isfinite(rt_map)]
    finite_map = finite_map[finite_map != 0]

    if finite_map.size == 0:
        map_vmin, map_vmax = -1, 1
    else:
        if symmetric:
            vmax = np.percentile(np.abs(finite_map), map_pct)
            map_vmin, map_vmax = -vmax, vmax
        else:
            map_vmin, map_vmax = np.percentile(
                finite_map,
                [100 - map_pct, map_pct]
            )

    # transparent mask
    if show_zero_as_transparent:
        map2d_plot = np.ma.masked_where(
            (~np.isfinite(map2d)) | (map2d == 0),
            map2d
        )
    else:
        map2d_plot = np.ma.masked_where(
            ~np.isfinite(map2d),
            map2d
        )

    plt.figure(figsize=(7, 7))
    plt.imshow(
        bg2d,
        cmap="gray",
        origin="lower",
        vmin=bg_vmin,
        vmax=bg_vmax
    )

    im = plt.imshow(
        map2d_plot,
        cmap="coolwarm",
        origin="lower",
        alpha=alpha,
        vmin=map_vmin,
        vmax=map_vmax
    )

    plt.colorbar(im, fraction=0.046, pad=0.04)
    plt.title(
        f"{SUBJECT}\n{map_name}\n"
        f"BG volume {volume}/{nt} | {label}"
    )
    plt.axis("off")
    plt.show()


interact(
    show_rapidtide_map,
    map_name=Dropdown(
        options=list(rt_map_dict.keys()),
        value=list(rt_map_dict.keys())[0],
        description="Map"
    ),
    volume=IntSlider(
        value=1,
        min=1,
        max=nt,
        step=1,
        description="BG vol"
    ),
    plane=Dropdown(
        options=["axial", "coronal", "sagittal"],
        value="axial",
        description="Plane"
    ),
    slice_idx=IntSlider(
        value=nz // 2,
        min=0,
        max=max(nx, ny, nz) - 1,
        step=1,
        description="Slice"
    ),
    alpha=FloatSlider(
        value=0.65,
        min=0.0,
        max=1.0,
        step=0.05,
        description="Alpha"
    ),
    bg_vmin_pct=FloatSlider(
        value=1,
        min=0,
        max=10,
        step=0.5,
        description="BG min%"
    ),
    bg_vmax_pct=FloatSlider(
        value=99,
        min=90,
        max=100,
        step=0.5,
        description="BG max%"
    ),
    map_pct=FloatSlider(
        value=98,
        min=80,
        max=100,
        step=0.5,
        description="Map pct"
    ),
    symmetric=Checkbox(
        value=True,
        description="Symmetric"
    ),
    show_zero_as_transparent=Checkbox(
        value=True,
        description="Zero transparent"
    )
)
```

```python
# DEFs

def shift_regressor_by_lag(reg, t, lag_sec, sign=-1):
    """
    sign=-1:
        predictor(t) = reg(t - lag)
        Positive lag means voxel response is delayed relative to regressor.

    sign=+1:
        predictor(t) = reg(t + lag)
        Use only if rapidtide sign convention appears reversed.
    """
    shifted_time = t + sign * lag_sec
    return np.interp(
        shifted_time,
        t,
        reg,
        left=np.nan,
        right=np.nan
    )


def fit_simple_glm(y, x, min_valid=100):
    """
    y: voxel percent signal change
    x: lag-corrected GM global percent regressor

    Returns:
        beta, intercept, r, r2, t_beta, n_valid, resid_std
    """
    valid = np.isfinite(y) & np.isfinite(x)

    if valid.sum() < min_valid:
        return np.nan, np.nan, np.nan, np.nan, np.nan, valid.sum(), np.nan

    yy = y[valid].astype(np.float64)
    xx = x[valid].astype(np.float64)

    x_mean = xx.mean()
    y_mean = yy.mean()

    xc = xx - x_mean
    yc = yy - y_mean

    ssx = np.sum(xc ** 2)
    ssy = np.sum(yc ** 2)

    if ssx <= 0 or ssy <= 0:
        return np.nan, np.nan, np.nan, np.nan, np.nan, valid.sum(), np.nan

    beta = np.sum(xc * yc) / ssx
    intercept = y_mean - beta * x_mean

    y_hat = intercept + beta * xx
    resid = yy - y_hat

    dof = valid.sum() - 2
    if dof > 0:
        resid_var = np.sum(resid ** 2) / dof
        resid_std = np.sqrt(resid_var)
        se_beta = np.sqrt(resid_var / ssx)
        t_beta = beta / se_beta if se_beta > 0 else np.nan
    else:
        resid_std = np.nan
        t_beta = np.nan

    r = np.sum(xc * yc) / np.sqrt(ssx * ssy)
    r2 = r ** 2

    return beta, intercept, r, r2, t_beta, valid.sum(), resid_std
```

```python
## Run voxelwise GLM

# Sign convention
# Default: positive rapidtide lag = voxel delayed, so use reg(t - lag)
LAG_SIGN = -1

MIN_VALID = int(nt * 0.70)

beta_map = np.full((nx, ny, nz), np.nan, dtype=np.float32)
intercept_map = np.full((nx, ny, nz), np.nan, dtype=np.float32)
r_map = np.full((nx, ny, nz), np.nan, dtype=np.float32)
r2_map = np.full((nx, ny, nz), np.nan, dtype=np.float32)
t_map = np.full((nx, ny, nz), np.nan, dtype=np.float32)
nvalid_map = np.zeros((nx, ny, nz), dtype=np.float32)
residstd_map = np.full((nx, ny, nz), np.nan, dtype=np.float32)

voxel_indices = np.array(np.where(stable_lag_mask)).T

print("Running GLM voxels:", len(voxel_indices))
print("MIN_VALID:", MIN_VALID)
print("LAG_SIGN:", LAG_SIGN)

for idx, (i, j, k) in enumerate(voxel_indices):
    if idx % 5000 == 0:
        print(f"{idx}/{len(voxel_indices)}")

    lag_sec = float(lag_map[i, j, k])

    y_raw = bold_data[i, j, k, :].astype(np.float64)

    y_mean = np.nanmean(y_raw)
    if not np.isfinite(y_mean) or y_mean == 0:
        continue

    # voxel percent signal change
    y_percent = 100.0 * (y_raw / y_mean - 1.0)

    # lag-corrected regressor
    x_shifted = shift_regressor_by_lag(
        gm_global_percent,
        t,
        lag_sec,
        sign=LAG_SIGN
    )

    beta, intercept, r, r2, t_beta, n_valid, resid_std = fit_simple_glm(
        y_percent,
        x_shifted,
        min_valid=MIN_VALID
    )

    beta_map[i, j, k] = beta
    intercept_map[i, j, k] = intercept
    r_map[i, j, k] = r
    r2_map[i, j, k] = r2
    t_map[i, j, k] = t_beta
    nvalid_map[i, j, k] = n_valid
    residstd_map[i, j, k] = resid_std

print("GLM finished.")
```

```python
def map_stats(path, mask_path=None):
    img = nib.load(str(path))
    data = np.squeeze(img.get_fdata(dtype=np.float32))

    if mask_path is not None:
        mask = np.squeeze(nib.load(str(mask_path)).get_fdata()) > 0
        vals = data[mask]
    else:
        vals = data.ravel()

    vals = vals[np.isfinite(vals)]
    vals = vals[vals != 0]

    print("file:", path.name)
    print("n:", vals.size)
    print("min/max:", np.min(vals), np.max(vals))
    print("percentiles:")
    for p in [0.1, 1, 2, 5, 25, 50, 75, 95, 98, 99, 99.9]:
        print(f"  p{p:>4}: {np.percentile(vals, p)}")

cvr_path = rt_result_dir / "Test_1a_desc-CVR_map.nii.gz"
corrfit_path = rt_result_dir / "Test_1a_desc-corrfit_mask.nii.gz"
processed_path = rt_result_dir / "Test_1a_desc-processed_mask.nii.gz"

map_stats(cvr_path, corrfit_path)
```
