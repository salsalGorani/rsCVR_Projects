```python
from pathlib import Path
import pandas as pd
```

```python
# DIRs

WORK_NAME = "M01_ds003768"


DATA_ROOT = Path("/media/bami/JHA/rsCVR/data/OpenNeuro/ds003768")

RESULTS_ROOT = DATA_ROOT.parent / "rsCVR_Results"
WORK_ROOT = RESULTS_ROOT / WORK_NAME

INDEX_DIR = WORK_ROOT / "index"
LOG_DIR = WORK_ROOT / "logs"
QC_DIR = WORK_ROOT / "qc"
RAPIDTIDE_DIR = WORK_ROOT / "rapidtide"
SPM_DIR = WORK_ROOT / "spm_preproc"
SUMMARY_DIR = WORK_ROOT / "summary"
MANIFEST_DIR = WORK_ROOT / "manifests"

for d in [RESULTS_ROOT, WORK_ROOT, INDEX_DIR, LOG_DIR, QC_DIR, RAPIDTIDE_DIR, SPM_DIR, SUMMARY_DIR, MANIFEST_DIR]:
    d.mkdir(parents=True, exist_ok=True)

print("DATA_ROOT    :", DATA_ROOT)
print("RESULTS_ROOT :", RESULTS_ROOT)
print("WORK_NAME    :", WORK_NAME)
print("WORK_ROOT    :", WORK_ROOT)
```

```python
# Load DataIndexing outputs

index_csv = INDEX_DIR / "subject_index.csv"
meta_csv = INDEX_DIR / "subject_meta.csv"
shape_csv = INDEX_DIR / "subject_shapes.csv"

df_index = pd.read_csv(index_csv)
df_meta = pd.read_csv(meta_csv)
df_shapes = pd.read_csv(shape_csv)

df_ok = df_index[df_index["complete_for_analysis"] == True].copy()

print("Loaded subject_index :", df_index.shape)
print("Loaded subject_meta  :", df_meta.shape)
print("Loaded subject_shapes:", df_shapes.shape)
print("Complete subjects    :", len(df_ok))
```

```python
# Subject-Lv output folder making

def make_subject_output_dirs(subject: str):
    spm_sub_root = SPM_DIR / subject
    rapid_sub_root = RAPIDTIDE_DIR / subject
    qc_sub_root = QC_DIR / subject

    dirs = {
        "spm_subject_root": spm_sub_root,
        "spm_anat_dir": spm_sub_root / "anat",
        "spm_run1_dir": spm_sub_root / "run-1",
        "spm_run2_dir": spm_sub_root / "run-2",

        "rapidtide_subject_root": rapid_sub_root,
        "rapidtide_run1_dir": rapid_sub_root / "run-1",
        "rapidtide_run2_dir": rapid_sub_root / "run-2",

        "qc_subject_root": qc_sub_root,
        "qc_run1_dir": qc_sub_root / "run-1",
        "qc_run2_dir": qc_sub_root / "run-2",
    }

    for d in dirs.values():
        d.mkdir(parents=True, exist_ok=True)

    return {k: str(v) for k, v in dirs.items()}

# Dir table
subject_dir_rows = []
for _, row in df_ok.iterrows():
    subject = row["subject"]
    outdirs = make_subject_output_dirs(subject)
    subject_dir_rows.append({
        "subject": subject,
        **outdirs
    })
df_subject_dirs = pd.DataFrame(subject_dir_rows)

print(df_subject_dirs.shape)
df_subject_dirs
```

```python
# Run-Lv output folder making

run_rows = []

for _, row in df_ok.iterrows():
    subject = row["subject"]
    subject_dirs = df_subject_dirs[df_subject_dirs["subject"] == subject].iloc[0]

    run_specs = [
        ("run-1", row["rest1_nii"], row["rest1_json"],
         subject_dirs["spm_run1_dir"], subject_dirs["rapidtide_run1_dir"], subject_dirs["qc_run1_dir"]),
        ("run-2", row["rest2_nii"], row["rest2_json"],
         subject_dirs["spm_run2_dir"], subject_dirs["rapidtide_run2_dir"], subject_dirs["qc_run2_dir"]),
    ]

    for run_label, bold_nii, bold_json, spm_run_dir, rapid_run_dir, qc_run_dir in run_specs:
        run_rows.append({
            "subject": subject,
            "run": run_label,

            "t1_nii": row["t1_nii"],
            "t1_json": row["t1_json"],

            "bold_nii": bold_nii,
            "bold_json": bold_json,

            "spm_subject_root": subject_dirs["spm_subject_root"],
            "spm_anat_dir": subject_dirs["spm_anat_dir"],
            "spm_run_dir": spm_run_dir,

            "rapidtide_subject_root": subject_dirs["rapidtide_subject_root"],
            "rapidtide_run_dir": rapid_run_dir,

            "qc_subject_root": subject_dirs["qc_subject_root"],
            "qc_run_dir": qc_run_dir,
        })

df_runs = pd.DataFrame(run_rows)

print(df_runs.shape)
df_runs
```

```python
# Exp. output DIRs
def add_expected_paths(df):
    df = df.copy()

    df["spm_preproc_bold"] = df.apply(
        lambda x: str(Path(x["spm_run_dir"]) / f"{x['subject']}_{x['run']}_preproc_bold.nii.gz"),
        axis=1
    )

    df["spm_motion_tsv"] = df.apply(
        lambda x: str(Path(x["spm_run_dir"]) / f"{x['subject']}_{x['run']}_motion.tsv"),
        axis=1
    )

    df["spm_brainmask"] = df.apply(
        lambda x: str(Path(x["spm_run_dir"]) / f"{x['subject']}_{x['run']}_brainmask.nii.gz"),
        axis=1
    )

    df["rapidtide_output_root"] = df.apply(
        lambda x: str(Path(x["rapidtide_run_dir"]) / f"{x['subject']}_{x['run']}"),
        axis=1
    )

    df["qc_summary_json"] = df.apply(
        lambda x: str(Path(x["qc_run_dir"]) / f"{x['subject']}_{x['run']}_qc_summary.json"),
        axis=1
    )

    return df

df_runs = add_expected_paths(df_runs)
df_runs
```

```python
# SAVE

subject_dirs_csv = MANIFEST_DIR / "subject_output_dirs.csv"
df_subject_dirs.to_csv(subject_dirs_csv, index=False)
print(subject_dirs_csv)

run_manifest_csv = MANIFEST_DIR / "run_manifest.csv"
df_runs.to_csv(run_manifest_csv, index=False)
print(run_manifest_csv)
```
