```python
def check_plane_shapes_raw(path, name=None):
    path = Path(path)
    img = nib.load(str(path))
    data = img.get_fdata(dtype=np.float32)

    if data.ndim == 4:
        data = data[..., 0]

    x, y, z = data.shape[:3]

    sag = data[x // 2, :, :]   # x fixed → y-z plane
    cor = data[:, y // 2, :]   # y fixed → x-z plane
    axi = data[:, :, z // 2]   # z fixed → x-y plane

    print(f"\n=== {name if name else path.name} ===")
    print("image shape :", data.shape)
    print("zooms       :", img.header.get_zooms()[:3])
    print("Sagittal slice shape:", sag.shape)
    print("Coronal  slice shape:", cor.shape)
    print("Axial    slice shape:", axi.shape)

check_plane_shapes_raw(T1_PATH, "3DT1 raw")
check_plane_shapes_raw(MEAN_BOLD, "meanBOLD raw")
check_plane_shapes_raw(BOLD_4D, "rsBOLD first volume raw")
```

```python
def check_plane_shapes_canonical(path, name=None):
    path = Path(path)
    img = nib.load(str(path))
    img_can = nib.as_closest_canonical(img)
    data = img_can.get_fdata(dtype=np.float32)

    if data.ndim == 4:
        data = data[..., 0]

    x, y, z = data.shape[:3]

    sag = data[x // 2, :, :]   # canonical x fixed
    cor = data[:, y // 2, :]   # canonical y fixed
    axi = data[:, :, z // 2]   # canonical z fixed

    print(f"\n=== {name if name else path.name} | canonical ===")
    print("original image shape :", nib.load(str(path)).shape)
    print("canonical image shape:", data.shape)
    print("canonical zooms      :", img_can.header.get_zooms()[:3])
    print("Sagittal slice shape :", sag.shape)
    print("Coronal  slice shape :", cor.shape)
    print("Axial    slice shape :", axi.shape)

check_plane_shapes_canonical(T1_PATH, "3DT1")
check_plane_shapes_canonical(MEAN_BOLD, "meanBOLD")
check_plane_shapes_canonical(BOLD_4D, "rsBOLD first volume")
```

```python
def get_world_corners(img):
    shape = img.shape[:3]
    corners = np.array([
        [0, 0, 0, 1],
        [shape[0]-1, 0, 0, 1],
        [0, shape[1]-1, 0, 1],
        [0, 0, shape[2]-1, 1],
        [shape[0]-1, shape[1]-1, 0, 1],
        [shape[0]-1, 0, shape[2]-1, 1],
        [0, shape[1]-1, shape[2]-1, 1],
        [shape[0]-1, shape[1]-1, shape[2]-1, 1],
    ])
    return (img.affine @ corners.T).T[:, :3]

t1_img = nib.load(str(T1_PATH))
mean_img = nib.load(str(MEAN_BOLD))

t1_axcodes = nib.aff2axcodes(t1_img.affine)
bold_axcodes = nib.aff2axcodes(mean_img.affine)

print("T1 shape      :", t1_img.shape)
print("T1 axcodes    :", t1_axcodes)
print("meanBOLD shape:", mean_img.shape)
print("meanBOLD axcodes:", bold_axcodes)

# BOLD world corners → T1 voxel coordinates
bold_world = get_world_corners(mean_img)
bold_world_h = np.c_[bold_world, np.ones(len(bold_world))]

bold_in_t1_vox = (np.linalg.inv(t1_img.affine) @ bold_world_h.T).T[:, :3]

print("\nBOLD FOV projected into T1 voxel coordinates")
print("min:", bold_in_t1_vox.min(axis=0))
print("max:", bold_in_t1_vox.max(axis=0))
```

```python
def plane_name_from_fixed_axis(axcodes, fixed_axis):
    fixed_code = axcodes[fixed_axis]

    if fixed_code in ("S", "I"):
        return "Axial"
    elif fixed_code in ("A", "P"):
        return "Coronal"
    elif fixed_code in ("R", "L"):
        return "Sagittal"
    else:
        return "Unknown"


def show_slices_along_axis(path, fixed_axis, idx_list=None, title=None):
    img = nib.load(str(path))
    data = img.get_fdata(dtype=np.float32)
    axcodes = nib.aff2axcodes(img.affine)

    n = data.shape[fixed_axis]

    if idx_list is None:
        idx_list = [
            int(n * 0.15),
            int(n * 0.30),
            int(n * 0.45),
            int(n * 0.60),
            int(n * 0.75),
            int(n * 0.90),
        ]

    plane_name = plane_name_from_fixed_axis(axcodes, fixed_axis)

    print("file:", path)
    print("shape:", data.shape)
    print("axcodes:", axcodes)
    print(f"fixed axis {fixed_axis} = {axcodes[fixed_axis]} direction")
    print(f"interpreted plane: {plane_name}")

    fig, axes = plt.subplots(
        1, len(idx_list),
        figsize=(3 * len(idx_list), 3),
        facecolor="black"
    )

    if len(idx_list) == 1:
        axes = [axes]

    for ax, idx in zip(axes, idx_list):
        slc = np.take(data, idx, axis=fixed_axis)

        ax.set_facecolor("black")
        ax.imshow(np.rot90(slc), cmap="gray", aspect="auto")
        ax.set_title(f"axis{fixed_axis}={idx}", color="white")
        ax.axis("off")

    fig.suptitle(
        title if title else f"{plane_name} view by fixing axis {fixed_axis}",
        color="white"
    )
    plt.tight_layout()
    plt.show()
```

```python
# Defs

def show_basic_info(path, name=None):
    path = Path(path)
    print("\n===", name if name else path.name, "===")
    print("path  :", path)
    print("exists:", path.exists())
    if path.exists():
        img = nib.load(str(path))
        print("shape :", img.shape)
        print("zooms :", img.header.get_zooms()[:3])
        print("affine:\n", img.affine)

def show_axis1_axial(path, idx_list=None, title=None):
    img = nib.load(str(path))
    data = img.get_fdata(dtype=np.float32)

    if data.ndim == 4:
        data = data[..., 0]

    axis = 1
    n = data.shape[axis]

    if idx_list is None:
        idx_list = [
            int(n * 0.15),
            int(n * 0.30),
            int(n * 0.45),
            int(n * 0.60),
            int(n * 0.75),
            int(n * 0.90),
        ]

    fig, axes = plt.subplots(
        1, len(idx_list),
        figsize=(3 * len(idx_list), 3),
        facecolor="black"
    )

    if len(idx_list) == 1:
        axes = [axes]

    for ax, idx in zip(axes, idx_list):
        slc = np.take(data, idx, axis=axis)

        ax.set_facecolor("black")
        ax.imshow(slc, cmap="gray", aspect="auto")
        ax.set_title(f"axis1={idx}", color="white")
        ax.axis("off")

    fig.suptitle(title if title else Path(path).name, color="white")
    plt.tight_layout()
    plt.show()

    print("shape:", img.shape)
    print("zooms:", img.header.get_zooms()[:3])

def show_same_grid_pair(img1_path, img2_path, title1="img1", title2="img2"):
    img1 = nib.load(str(img1_path))
    img2 = nib.load(str(img2_path))

    d1 = img1.get_fdata(dtype=np.float32)
    d2 = img2.get_fdata(dtype=np.float32)

    if d1.ndim == 4:
        d1 = d1[..., 0]
    if d2.ndim == 4:
        d2 = d2[..., 0]

    if d1.shape != d2.shape:
        raise ValueError(f"Shape mismatch: {d1.shape} vs {d2.shape}")

    x, y, z = d1.shape

    views = [
        ("axis0 fixed", d1[x//2, :, :], d2[x//2, :, :]),
        ("axis1 fixed", d1[:, y//2, :], d2[:, y//2, :]),
        ("axis2 fixed", d1[:, :, z//2], d2[:, :, z//2]),
    ]

    fig, axes = plt.subplots(2, 3, figsize=(12, 7), facecolor="black")

    for i, (name, s1, s2) in enumerate(views):
        axes[0, i].imshow(np.rot90(s1), cmap="gray", aspect="auto")
        axes[0, i].set_title(f"{title1} | {name}", color="white")
        axes[0, i].axis("off")

        axes[1, i].imshow(np.rot90(s2), cmap="gray", aspect="auto")
        axes[1, i].set_title(f"{title2} | {name}", color="white")
        axes[1, i].axis("off")

    for ax in axes.ravel():
        ax.set_facecolor("black")

    plt.tight_layout()
    plt.show()

def overlay_contour_same_grid(base_path, overlay_path, title=None, pct=40):
    base_img = nib.load(str(base_path))
    ov_img = nib.load(str(overlay_path))

    base = base_img.get_fdata(dtype=np.float32)
    ov = ov_img.get_fdata(dtype=np.float32)

    if base.ndim == 4:
        base = base[..., 0]
    if ov.ndim == 4:
        ov = ov[..., 0]

    if base.shape != ov.shape:
        raise ValueError(f"Shape mismatch: base={base.shape}, overlay={ov.shape}")

    x, y, z = base.shape

    vals = ov[np.isfinite(ov) & (ov > 0)]
    ov_thr = np.percentile(vals, pct)

    views = [
        ("axis0 fixed", base[x//2, :, :], ov[x//2, :, :]),
        ("axis1 fixed", base[:, y//2, :], ov[:, y//2, :]),
        ("axis2 fixed", base[:, :, z//2], ov[:, :, z//2]),
    ]

    fig, axes = plt.subplots(1, 3, figsize=(12, 4), facecolor="black")

    for ax, (name, b, o) in zip(axes, views):
        ax.set_facecolor("black")
        ax.imshow(np.rot90(b), cmap="gray", aspect="auto")
        ax.contour(np.rot90(o > ov_thr), levels=[0.5], colors="red", linewidths=1)
        ax.set_title(name, color="white")
        ax.axis("off")

    fig.suptitle(title if title else f"{Path(base_path).name} + {Path(overlay_path).name}", color="white")
    plt.tight_layout()
    plt.show()
```

```python
def brain_center_world(path, q=60):
    img = nib.load(str(path))
    data = img.get_fdata(dtype=np.float32)

    if data.ndim == 4:
        data = data[..., 0]

    valid = np.isfinite(data) & (data > 0)
    thr = np.percentile(data[valid], q)
    mask = valid & (data > thr)

    coords = np.argwhere(mask)
    centre_vox = coords.mean(axis=0)

    centre_world = nib.affines.apply_affine(img.affine, centre_vox)

    return centre_vox, centre_world


mean_vox, mean_world = brain_center_world(MEAN_BOLD, q=60)
t1_vox, t1_world = brain_center_world(T1_FOR_SEG, q=60)

delta_world = mean_world - t1_world

print("meanBOLD centre voxel:", mean_vox)
print("meanBOLD centre world:", mean_world)

print("T1_FOR_SEG centre voxel:", t1_vox)
print("T1_FOR_SEG centre world:", t1_world)

print("world shift needed:", delta_world)
```

```python
def show_axis1_axial_no_rotation(path, idx_list=None, title=None, vmin=None, vmax=None):
    img = nib.load(str(path))
    data = img.get_fdata(dtype=np.float32)

    axis = 1
    n = data.shape[axis]

    if idx_list is None:
        idx_list = [
            int(n * 0.15),
            int(n * 0.30),
            int(n * 0.45),
            int(n * 0.60),
            int(n * 0.75),
            int(n * 0.90),
        ]

    fig, axes = plt.subplots(1, len(idx_list), figsize=(3 * len(idx_list), 3), facecolor="black")
    if len(idx_list) == 1:
        axes = [axes]

    for ax, idx in zip(axes, idx_list):
        slc = np.take(data, idx, axis=axis)
        ax.set_facecolor("black")
        ax.imshow(slc, cmap="gray", aspect="auto", vmin=vmin, vmax=vmax)
        ax.set_title(f"axis1={idx}", color="white")
        ax.axis("off")

    fig.suptitle(title if title else Path(path).name, color="white")
    plt.tight_layout()
    plt.show()

    print("shape:", img.shape)
    print("zooms:", img.header.get_zooms()[:3])

show_axis1_axial_no_rotation(T1_FOR_SEG, title="T1_FOR_SEG | axis1 axial")

show_axis1_axial_no_rotation(c1_native, title="native c1 GM | axis1 axial", vmin=0, vmax=1)
show_axis1_axial_no_rotation(c2_native, title="native c2 WM | axis1 axial", vmin=0, vmax=1)
show_axis1_axial_no_rotation(c3_native, title="native c3 CSF | axis1 axial", vmin=0, vmax=1)
```

```python
def overlay_prob_on_axis1_t1(t1_path, prob_path, idx_list=None, prob_thr=0.5, title=None):
    t1_img = nib.load(str(t1_path))
    prob_img = nib.load(str(prob_path))

    t1 = t1_img.get_fdata(dtype=np.float32)
    prob = prob_img.get_fdata(dtype=np.float32)

    if t1.shape != prob.shape:
        raise ValueError(f"Shape mismatch: T1={t1.shape}, prob={prob.shape}")

    axis = 1
    n = t1.shape[axis]

    if idx_list is None:
        idx_list = [
            int(n * 0.15),
            int(n * 0.30),
            int(n * 0.45),
            int(n * 0.60),
            int(n * 0.75),
            int(n * 0.90),
        ]

    fig, axes = plt.subplots(1, len(idx_list), figsize=(3 * len(idx_list), 3), facecolor="black")
    if len(idx_list) == 1:
        axes = [axes]

    for ax, idx in zip(axes, idx_list):
        t1_slc = np.take(t1, idx, axis=axis)
        prob_slc = np.take(prob, idx, axis=axis)

        ax.set_facecolor("black")
        ax.imshow(t1_slc, cmap="gray", aspect="auto")
        ax.contour(prob_slc > prob_thr, levels=[0.5], colors="red", linewidths=1)
        ax.set_title(f"axis1={idx}", color="white")
        ax.axis("off")

    fig.suptitle(title if title else Path(prob_path).name, color="white")
    plt.tight_layout()
    plt.show()

overlay_prob_on_axis1_t1(
    T1_FOR_SEG,
    c1_native,
    prob_thr=0.5,
    title="T1_FOR_SEG + native c1 GM contour"
)

overlay_prob_on_axis1_t1(
    T1_FOR_SEG,
    c2_native,
    prob_thr=0.5,
    title="T1_FOR_SEG + native c2 WM contour"
)

overlay_prob_on_axis1_t1(
    T1_FOR_SEG,
    c3_native,
    prob_thr=0.5,
    title="T1_FOR_SEG + native c3 CSF contour"
)
```

```python
# Compare meanBOLD <--> resliced T1

def compare_same_grid(img1_path, img2_path, title1="img1", title2="img2"):
    img1 = nib.load(str(img1_path))
    img2 = nib.load(str(img2_path))

    d1 = img1.get_fdata(dtype=np.float32)
    d2 = img2.get_fdata(dtype=np.float32)

    if d1.ndim == 4:
        d1 = d1[..., 0]
    if d2.ndim == 4:
        d2 = d2[..., 0]

    if d1.shape != d2.shape:
        raise ValueError(f"Shape mismatch: {d1.shape} vs {d2.shape}")

    x, y, z = d1.shape

    views = [
        ("Sagittal", d1[x//2, :, :], d2[x//2, :, :]),
        ("Coronal",  d1[:, y//2, :], d2[:, y//2, :]),
        ("Axial",    d1[:, :, z//2], d2[:, :, z//2]),
    ]

    fig, axes = plt.subplots(2, 3, figsize=(12, 7), facecolor="black")

    for i, (name, s1, s2) in enumerate(views):
        axes[0, i].imshow(np.rot90(s1), cmap="gray", aspect="auto")
        axes[0, i].set_title(f"{title1} {name}", color="white")
        axes[0, i].axis("off")

        axes[1, i].imshow(np.rot90(s2), cmap="gray", aspect="auto")
        axes[1, i].set_title(f"{title2} {name}", color="white")
        axes[1, i].axis("off")

    for ax in axes.ravel():
        ax.set_facecolor("black")

    plt.tight_layout()
    plt.show()

compare_same_grid(
    MEAN_BOLD,
    rT1,
    title1="meanBOLD",
    title2="resliced T1"
)
```

```python
# Brain mask creation
thr = 0.5

gm = nib.load(str(c1)).get_fdata(dtype=np.float32)
wm = nib.load(str(c2)).get_fdata(dtype=np.float32)
csf = nib.load(str(c3)).get_fdata(dtype=np.float32)

brain_prob = gm + wm + csf
mask = (brain_prob > thr).astype(np.uint8)

mean_img = nib.load(str(MEAN_BOLD))

if mask.shape != mean_img.shape:
    raise ValueError(f"Shape mismatch: mask={mask.shape}, meanBOLD={mean_img.shape}")

nib.save(
    nib.Nifti1Image(mask, mean_img.affine, mean_img.header),
    str(MASK_PATH)
)

print("Saved BOLD-space mask:", MASK_PATH)
print("mask shape:", mask.shape)
print("mask voxels:", mask.sum())
print("mask ratio :", mask.mean())
```

```python
# MASK confirm
def quick_check_mean_mask(mean_path, mask_path):
    mean_img = nib.load(str(mean_path))
    mean_data = mean_img.get_fdata(dtype=np.float32)

    mask_img = nib.load(str(mask_path))
    mask_data = mask_img.get_fdata() > 0

    x, y, z = mean_data.shape

    slices = [
        ("Sagittal", mean_data[x//2, :, :], mask_data[x//2, :, :]),
        ("Coronal", mean_data[:, y//2, :], mask_data[:, y//2, :]),
        ("Axial", mean_data[:, :, z//2], mask_data[:, :, z//2]),
    ]

    fig, axes = plt.subplots(2, 3, figsize=(12, 7), facecolor="black")

    for i, (name, mean_slc, mask_slc) in enumerate(slices):
        axes[0, i].imshow(np.rot90(mean_slc), cmap="gray", aspect="auto")
        axes[0, i].set_title(f"meanBOLD {name}", color="white")
        axes[0, i].axis("off")

        axes[1, i].imshow(np.rot90(mask_slc), cmap="gray", aspect="auto")
        axes[1, i].set_title(f"mask {name}", color="white")
        axes[1, i].axis("off")

    for ax in axes.ravel():
        ax.set_facecolor("black")

    plt.tight_layout()
    plt.show()

quick_check_mean_mask(MEAN_BOLD, MASK_PATH)
```
